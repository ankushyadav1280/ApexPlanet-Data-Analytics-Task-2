```python
import pandas as pd

df = pd.read_excel("cleaned_sales_dataset (1).xlsx")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python
# 1. Dataset size
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

# 2. Column names
print("\nColumn Names:")
print(df.columns.tolist())

# 3. Data types and non-null values
print("\nDataset Information:")
df.info()

# 4. Statistical summary
print("\nStatistical Summary:")
display(df.describe(include="all"))

# 5. Missing values
print("\nMissing Values:")
display(df.isnull().sum())

# 6. Duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# Overall business metrics
total_revenue = df["Total_Sales"].sum()
total_quantity = df["Quantity"].sum()
average_order_value = df["Total_Sales"].mean()
unique_customers = df["Customer_ID"].nunique()
unique_products = df["Product"].nunique()

print(f"Total Sales: ₹{total_revenue:,.2f}")
print(f"Total Quantity Sold: {total_quantity:,}")
print(f"Average Transaction Sales: ₹{average_order_value:,.2f}")
print(f"Unique Customers: {unique_customers:,}")
print(f"Unique Products: {unique_products:,}")
```

```python
category_sales = (
    df.groupby("Category")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

display(category_sales)

category_sales.plot(kind="bar", figsize=(8, 5))
plt.title("Total Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Convert Order_Date into datetime
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

# Group sales by month
monthly_sales = (
    df.groupby(df["Order_Date"].dt.to_period("M"))["Total_Sales"]
    .sum()
)

monthly_sales.index = monthly_sales.index.astype(str)

print("Monthly Sales:")
display(monthly_sales)

# Plot monthly sales
plt.figure(figsize=(12, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Top 5 products by total sales
top_products = (
    df.groupby("Product")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

print("Top 5 Products by Sales:")
display(top_products)

# Bar chart
plt.figure(figsize=(10, 5))
top_products.sort_values().plot(kind="barh")
plt.title("Top 5 Products by Total Sales")
plt.xlabel("Total Sales (₹)")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

```python
city_sales = (
    df.groupby("City")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Top 5 Cities by Sales:")
display(city_sales.head(5))

plt.figure(figsize=(10, 5))
city_sales.head(5).sort_values().plot(kind="barh")
plt.title("Top 5 Cities by Sales")
plt.xlabel("Total Sales (₹)")
plt.ylabel("City")
plt.tight_layout()
plt.show()
```

```python
top_quantity = (
    df.groupby("Product")["Quantity"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

print("Top 5 Products by Quantity Sold:")
display(top_quantity)

plt.figure(figsize=(10, 5))
top_quantity.sort_values().plot(kind="barh")
plt.title("Top 5 Products by Quantity Sold")
plt.xlabel("Total Quantity Sold")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

```python
# Numerical columns summary
print("NUMERICAL SUMMARY")
display(df.describe())

# Categorical columns summary
print("CATEGORICAL SUMMARY")
display(df.describe(include="object"))

# Check missing values
print("MISSING VALUES")
display(df.isnull().sum())

# Check duplicates
print("DUPLICATE ROWS:", df.duplicated().sum())
```

```python
import sqlite3
import pandas as pd

# Create an in-memory SQL database
conn = sqlite3.connect(":memory:")

# Load the cleaned dataset into SQL
df.to_sql("sales", conn, if_exists="replace", index=False)

print("Dataset loaded into SQL successfully!")
```

```python
query1 = """
SELECT Product,
       ROUND(SUM(Total_Sales), 2) AS Revenue
FROM sales
GROUP BY Product
ORDER BY Revenue DESC
LIMIT 5;
"""

display(pd.read_sql_query(query1, conn))
```

```python
query2 = """
SELECT Category,
       ROUND(SUM(Total_Sales), 2) AS Revenue
FROM sales
GROUP BY Category
ORDER BY Revenue DESC;
"""

display(pd.read_sql_query(query2, conn))
```

```python
query3 = """
SELECT City,
       ROUND(SUM(Total_Sales), 2) AS Revenue
FROM sales
GROUP BY City
ORDER BY Revenue DESC
LIMIT 5;
"""

display(pd.read_sql_query(query3, conn))
```

```python
query4 = """
SELECT strftime('%Y-%m', Order_Date) AS Month,
       ROUND(SUM(Total_Sales), 2) AS Monthly_Sales
FROM sales
GROUP BY strftime('%Y-%m', Order_Date)
ORDER BY Month;
"""

display(pd.read_sql_query(query4, conn))
```

```python
query5 = """
SELECT Product,
       SUM(Quantity) AS Total_Quantity
FROM sales
GROUP BY Product
ORDER BY Total_Quantity DESC
LIMIT 5;
"""

display(pd.read_sql_query(query5, conn))
```

```python
query6 = """
SELECT Gender,
       COUNT(*) AS Transactions,
       ROUND(SUM(Total_Sales), 2) AS Revenue
FROM sales
GROUP BY Gender
ORDER BY Revenue DESC;
"""

display(pd.read_sql_query(query6, conn))
```

```python
query7 = """
SELECT
    CASE
        WHEN Age < 25 THEN 'Under 25'
        WHEN Age < 36 THEN '25-35'
        WHEN Age < 46 THEN '36-45'
        WHEN Age < 61 THEN '46-60'
        ELSE 'Above 60'
    END AS Age_Group,
    COUNT(*) AS Transactions,
    ROUND(SUM(Total_Sales), 2) AS Revenue
FROM sales
WHERE Age IS NOT NULL
GROUP BY Age_Group
ORDER BY Revenue DESC;
"""

display(pd.read_sql_query(query7, conn))
```

```python
results = {
    "Top_5_Products": pd.read_sql_query(query1, conn),
    "Category_Revenue": pd.read_sql_query(query2, conn),
    "Top_5_Cities": pd.read_sql_query(query3, conn),
    "Monthly_Sales": pd.read_sql_query(query4, conn),
    "Top_5_Quantity": pd.read_sql_query(query5, conn),
    "Gender_Revenue": pd.read_sql_query(query6, conn),
    "Age_Group_Revenue": pd.read_sql_query(query7, conn)
}

with pd.ExcelWriter("SQL_Query_Results.xlsx") as writer:
    for sheet_name, result in results.items():
        result.to_excel(writer, sheet_name=sheet_name, index=False)

print("SQL results saved to SQL_Query_Results.xlsx")
```

```python
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[0, 24, 35, 45, 60, float("inf")],
    labels=["Under 25", "25-35", "36-45", "46-60", "Above 60"]
)

age_category_sales = df.pivot_table(
    index="Age_Group",
    columns="Category",
    values="Total_Sales",
    aggfunc="sum",
    observed=False
)

display(age_category_sales)

plt.figure(figsize=(10, 6))
sns.heatmap(age_category_sales, annot=True, fmt=".0f")
plt.title("Sales by Age Group and Category")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# Prepare KPI summary
kpis = pd.DataFrame({
    "KPI": [
        "Total Sales",
        "Total Quantity Sold",
        "Unique Customers",
        "Unique Products",
        "Total Transactions"
    ],
    "Value": [
        df["Total_Sales"].sum(),
        df["Quantity"].sum(),
        df["Customer_ID"].nunique(),
        df["Product"].nunique(),
        len(df)
    ]
})

# Prepare chart data
category_data = (
    df.groupby("Category", as_index=False)["Total_Sales"]
    .sum()
    .sort_values("Total_Sales", ascending=False)
)

city_data = (
    df.groupby("City", as_index=False)["Total_Sales"]
    .sum()
    .sort_values("Total_Sales", ascending=False)
    .head(5)
)

monthly_data = (
    df.groupby(df["Order_Date"].dt.to_period("M"))["Total_Sales"]
    .sum()
    .reset_index()
)
monthly_data["Order_Date"] = monthly_data["Order_Date"].astype(str)

with pd.ExcelWriter("Dashboard_Data.xlsx") as writer:
    kpis.to_excel(writer, sheet_name="KPIs", index=False)
    category_data.to_excel(writer, sheet_name="Category Sales", index=False)
    city_data.to_excel(writer, sheet_name="Top Cities", index=False)
    monthly_data.to_excel(writer, sheet_name="Monthly Sales", index=False)

print("Dashboard_Data.xlsx created successfully!")
display(kpis)
```

```python

# Create EDA Report as a PDF
# Install reportlab if required: %pip install reportlab

from reportlab.lib.pagesizes import A4
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle
from reportlab.lib import colors
from reportlab.lib.styles import getSampleStyleSheet
from reportlab.lib.units import inch

styles = getSampleStyleSheet()
story = []

story.append(Paragraph("Exploratory Data Analysis Report", styles["Title"]))
story.append(Spacer(1, 12))

story.append(Paragraph("1. Dataset Overview", styles["Heading2"]))
story.append(Paragraph(
    f"Total records: {len(df)}<br/>"
    f"Total columns: {len(df.columns)}<br/>"
    f"Total sales: Rs. {df['Total_Sales'].sum():,.2f}",
    styles["BodyText"]
))
story.append(Spacer(1, 12))

story.append(Paragraph("2. Category-wise Sales", styles["Heading2"]))
category = df.groupby("Category")["Total_Sales"].sum().sort_values(ascending=False)
category_data = [["Category", "Total Sales (Rs.)"]]
category_data += [[str(k), f"{v:,.2f}"] for k, v in category.items()]

t1 = Table(category_data, repeatRows=1)
t1.setStyle(TableStyle([
    ("BACKGROUND", (0, 0), (-1, 0), colors.lightgrey),
    ("GRID", (0, 0), (-1, -1), 0.5, colors.grey),
    ("PADDING", (0, 0), (-1, -1), 6),
]))
story.append(t1)
story.append(Spacer(1, 12))

story.append(Paragraph("3. Top 5 Cities by Sales", styles["Heading2"]))
cities = df.groupby("City")["Total_Sales"].sum().sort_values(ascending=False).head(5)
city_data = [["City", "Total Sales (Rs.)"]]
city_data += [[str(k), f"{v:,.2f}"] for k, v in cities.items()]

t2 = Table(city_data, repeatRows=1)
t2.setStyle(TableStyle([
    ("BACKGROUND", (0, 0), (-1, 0), colors.lightgrey),
    ("GRID", (0, 0), (-1, -1), 0.5, colors.grey),
    ("PADDING", (0, 0), (-1, -1), 6),
]))
story.append(t2)
story.append(Spacer(1, 12))

story.append(Paragraph("4. Correlation Analysis", styles["Heading2"]))
corr = df[["Age", "Quantity", "Unit_Price", "Total_Sales"]].corr().round(2)
corr_data = [["Variable"] + list(corr.columns)]
for index, row in corr.iterrows():
    corr_data.append([str(index)] + [f"{x:.2f}" for x in row])

t3 = Table(corr_data, repeatRows=1)
t3.setStyle(TableStyle([
    ("BACKGROUND", (0, 0), (-1, 0), colors.lightgrey),
    ("GRID", (0, 0), (-1, -1), 0.5, colors.grey),
    ("PADDING", (0, 0), (-1, -1), 5),
]))
story.append(t3)
story.append(Spacer(1, 12))

story.append(Paragraph(
    "Key finding: Electronics is the highest-revenue category in this dataset. "
    "Quantity and Unit Price have positive correlations with Total Sales. "
    "Correlation does not establish causation.",
    styles["BodyText"]
))

doc = SimpleDocTemplate(
    "EDA_Report.pdf",
    pagesize=A4,
    rightMargin=0.6*inch,
    leftMargin=0.6*inch,
    topMargin=0.6*inch,
    bottomMargin=0.6*inch
)
doc.build(story)

print("EDA_Report.pdf created successfully!")
```
